**Theory Questions**

1. What are data structures, and why are they important?
 - Data Structures are organized ways to store manage, and retrieve data efficiently. They are important because they allow programs to handle large amount of data, optimize memory usage and improve algorithm performance for searching, sorting and manipulation.

2. Explain the difference between mutable and immutable data types with examples
 - Mutable: Can be changed after creation. Example-list, dict, set
 - Immutable: Cannot be changed after creation. Example-tuple, string, int

3. What are the main differences between lists and tuples in Python
 - List: It is mutable and we can change it frequently but it can can be slower for iteration, syntax for list is [1,2,3]
 - Tuple: It is immutable and it is fixed i.e, which cannot be changed and it is faster in performance, syntax for tuple is (1,2,3)

4. Describe how dictionaries store data
 - Dictionaries store data as key value pairs. Each key is unique and maps to a value. Python uses a hash table internally for fast access and retrieval of values by keys.

5. Why might you use a set instead of a list in Python
 - Sets automatically remove duplicates
 - Faster membership testing
 - Useful when order does not matter

6. What is a string in Python, and how is it different from a list
 - A string is an immutable sequence of characters
 - Unlike a list, elements in astring cannot be changed after creation
 - Strings support concatenation and slicing but not item assignment

7. How do tuples ensure data integrity in Python
 - Since tuples are immutable their content cannot be accidentally changed ensuring data consistency and integrity

8. What is a hash table, and how does it relate to dictionaries in Python
 -  A hash table stores key value pairs using hash function to map keys to indicate in an hash function to map keys to indicies in an internal array. Python dictionaries use hash tables for fast O(1) aerage time access to values

9. Can lists contain different data types in Python
 - Yes, lists can contain heterogenous elements

10. Explain why strings are immutable in Python
 - Strings are immutable because strings prevents accidental changes
 - They allow python to optimize memory usage and caching
 - Immutable objects are hashable, so they can be used as dictionaries keys

11. What advantages do dictionaries offer over lists for certain tasks
 - Fast lookups using keys (O(1) average)
 - Easier to represent structured data (key: value)
 - No need to search sequentially as in lists

12. Describe a scenario where using a tuple would be preferable over a list
  - Storing coordinates (x, y), RGB values, or database records where values should not change

13. How do sets handle duplicate values in Python
 - Sets automatically remove duplicates

14. How does the “in” keyword work differently for lists and dictionaries
 - Lists: Checks if a value exists
 - Dictionaries: Checks if a key exists, not the value

15. Can you modify the elements of a tuple? Explain why or why not
 - No, tuples are immutable, so elements cannot be changed once defined. You must create a new tuple to “modify” it

16. What is a nested dictionary, and give an example of its use case
 - A dictionary that contains another dictionary as a value
 - Use case: Storing complex structured data.

17. Describe the time complexity of accessing elements in a dictionary
 - Average case: O(1) using the key
 - Worst case: O(n) if hash collisions occur

18. In what situations are lists preferred over dictionaries
 - Maintaining ordered collections
 - When duplicate values are allowed
 - Sequential access is needed

19. Why are dictionaries considered unordered, and how does that affect data retrieval
 - Keys in a dictionary do not have a guaranteed order
 - Accessing by key is still O(1) but cannot rely on index positions

20. Explain the difference between a list and a dictionary in terms of data retrieval
 - List: Access by index, O(n) for searching value
 - Dictionary: Access by key, O(1) on average
 - Dictionaries are faster for key-based lookup; lists are better for ordered sequences
